### Core Causal Finding: Observational vs. Interventional Discovery
1. **Colliders are Solved (SHD = 0.0)**:
   - Topologies 2, 3, and 6 achieved **perfect structural discovery (SHD 0.0)**.
   - In causal DAG theory, Colliders ($X_i \to X_k \leftarrow X_j$) possess a unique **unshielded collider signature** in observational data ($X_i$ and $X_j$ are marginally independent but conditionally dependent given $X_k$). The agent's Dual-Head Graph Scorer successfully learned to identify this signature from the covariance matrix.
2. **Chains and Forks Fail (SHD = 2.0 to 4.0)**:
   - Topologies 0, 1, 4, and 5 failed with high SHD penalties.
   - Chains ($Z_1 \to X_1 \to X_2$) and Forks ($Z_1 \leftarrow X_1 \to X_2$) belong to the **same Markov Equivalence Class**. Their observational covariance matrices are statistically indistinguishable. Discovering their orientation requires **active interventions**.

## 4. Root Cause Bug: Action Target Mask Misalignment
Why did the agent fail to intervene to break Markov Equivalence for Chains/Forks?

### Diagnostic Evidence:
Examining the step-by-step actions for Agent 0 (Agent 1) in Topology 0:
- Agent 0 repeatedly chose `cat: 0` (Local Intervention) with `target: 2` (Node 2 / X2).
- **The Problem**: Node 2 is $X_2$, which belongs exclusively to Agent 2! Agent 1 does not own Node 2.
- **The Bug in `train.py` / `ppo_agent.py`**: `train.py` passed the **Observation Mask** (`[1, 1, 1, 0]`) into `mask_invalid_targets()`. The actor network saw a `1` at index 2 and permitted `target: 2` for `LOCAL_INTERVENTION`.
- **Environment Rejection**: In `evaluator_env.py`, the environment checked `self.agent_masks` (the **Ownership Mask** `[1, 1, 0, 0]`). Seeing that Agent 1 does not own Node 2, the environment silently rejected the action, resulting in a **NO-OP**.

### Implemented Code Fix:
We refactored `mask_invalid_targets` to take two explicit masks:
1. `local_ownership_mask` (`[1, 1, 0, 0]` for A1, `[0, 0, 1, 1]` for A2) -> strictly enforces local ownership for `LOCAL_INTERVENTION`.
2. `peer_boundary_mask` (`[0, 1, 1, 0]`) -> strictly enforces boundary node selection for `PEER_REQUEST`.
