### Theoretical Domain Explanation (Causal DAG & Markov Equivalence):
- **Why Colliders (Topologies 2, 3, 6) achieve Perfect Discovery (SHD = 0.0):**
  - Colliders ($X_i \to X_k \leftarrow X_j$) possess an unshielded collider structure. In purely observational covariance matrices, $X_i$ and $X_j$ are marginally independent ($Cov(X_i, X_j) = 0$) but become conditionally dependent given $X_k$. The shared Edge Scorer MLP in the Graph Head easily detects this unique observational signature and correctly outputs 0.0 SHD.
- **Why Chains and Forks (Topologies 0, 1, 4, 5) Fail (SHD = 2.0 – 3.0):**
  - Chains ($Z_1 \to X_1 \to X_2 \to Z_2$) and Forks ($Z_1 \leftarrow X_1 \to X_2 \to Z_2$) belong to the exact same **Markov Equivalence Class**. Their observational covariance matrices are mathematically identical. Without **active interventional shifts**, no machine learning model can distinguish direct edge directions from observational data alone.

## 4. Root Cause Diagnostic: Target Mask Misalignment Bug
### Empirical Evidence from Trace Action Logs:
In the action pattern column above, observe Agent 0's actions: `C0T2, C0T2, C0T2, C0T2, C0T2`.
- `C0` = Action Category 0 (`LOCAL_INTERVENTION`).
- `T2` = Action Target 2 ($X_2$, Agent 2's boundary node).

### Mechanics of the Failure Mode:
1. **The Mask Bug in `train.py` / `ppo_agent.py`:**
   - `train.py` passed the agent's **Observation Mask** (`[1, 1, 1, 0]`) into `mask_invalid_targets()`.
   - Because $X_2$ (index 2) is observed by Agent 1, the action masking code permitted `target=2` for Category 0 (`LOCAL_INTERVENTION`).
2. **Environment Action Rejection in `evaluator_env.py`:**
   - When the action reached `evaluator_env.py`, the environment checked `self.agent_masks[k, target]` (the **Ownership Mask** `[1, 1, 0, 0]`).
   - Because Agent 1 does not *own* Node 2, `evaluator_env.py` rejected the intervention and spent 0 budget, effectively forcing a **NO-OP**.
3. **Result:** The agent intended to intervene on $X_2$, but the environment silently treated it as a NO-OP. As a result, no interventional data was generated, preventing the agent from breaking Markov Equivalence for Chains and Forks!

### Implemented Refactoring Fix:
We refactored `mask_invalid_targets(cat_action, target_logits, local_ownership_mask, peer_boundary_mask)` so that `LOCAL_INTERVENTION` is strictly masked by `local_ownership_mask` (`[1, 1, 0, 0]` for A1, `[0, 0, 1, 1]` for A2), guaranteeing agents can only intervene on nodes they actually own.

````python
import json
import matplotlib.pyplot as plt
import numpy as np

# Load exact history json
with open('../scratch/wandb_run_05n0rr9e_history.json', 'r') as f:
    history = json.load(f)

episodes = [h['train/episode'] for h in history if 'train/episode' in h]
rewards = [h['train/episode_reward'] for h in history if 'train/episode_reward' in h]
actor_loss = [h['train/actor_loss'] for h in history if 'train/actor_loss' in h]
critic_loss = [h['train/critic_loss'] for h in history if 'train/critic_loss' in h]
graph_loss = [h['train/graph_loss'] for h in history if 'train/graph_loss' in h]
entropy = [h['train/entropy'] for h in history if 'train/entropy' in h]
eval_shd = [h['eval/shd'] for h in history if 'eval/shd' in h]
eval_f1 = [h['eval/f1'] for h in history if 'eval/f1' in h]

fig, axs = plt.subplots(2, 3, figsize=(18, 10))

axs[0, 0].plot(episodes, rewards, color='crimson', linewidth=2)
axs[0, 0].set_title('Episode Reward Trajectory')
axs[0, 0].set_xlabel('Episode')
axs[0, 0].set_ylabel('Reward')
axs[0, 0].grid(True, alpha=0.3)

axs[0, 1].plot(episodes, actor_loss, label='Actor Loss', color='blue')
axs[0, 1].plot(episodes, critic_loss, label='Critic Loss', color='orange')
axs[0, 1].plot(episodes, graph_loss, label='Graph BCE Loss', color='green')
axs[0, 1].set_title('Loss Curves (Actor, Critic, Graph)')
axs[0, 1].set_xlabel('Episode')
axs[0, 1].set_ylabel('Loss')
axs[0, 1].legend()
axs[0, 1].grid(True, alpha=0.3)

axs[0, 2].plot(episodes, entropy, color='purple', linewidth=2)
axs[0, 2].set_title('Policy Entropy')
axs[0, 2].set_xlabel('Episode')
axs[0, 2].set_ylabel('Entropy')
axs[0, 2].grid(True, alpha=0.3)

axs[1, 0].plot(episodes, eval_shd, color='navy', linewidth=2)
axs[1, 0].set_title('Evaluation SHD Trajectory')
axs[1, 0].set_xlabel('Episode')
axs[1, 0].set_ylabel('SHD (Lower is better)')
axs[1, 0].grid(True, alpha=0.3)

axs[1, 1].plot(episodes, eval_f1, color='forestgreen', linewidth=2)
axs[1, 1].set_title('Evaluation F1 Score Trajectory')
axs[1, 1].set_xlabel('Episode')
axs[1, 1].set_ylabel('F1 Score (Higher is better)')
axs[1, 1].grid(True, alpha=0.3)

a0_b = [h['agent_0_budget'] for h in history if 'agent_0_budget' in h]
a1_b = [h['agent_1_budget'] for h in history if 'agent_1_budget' in h]
axs[1, 2].plot(episodes, a0_b, label='Agent 0 Budget', color='darkcyan')
axs[1, 2].plot(episodes, a1_b, label='Agent 1 Budget', color='magenta', linestyle='--')
axs[1, 2].set_title('Remaining Agent Budgets')
axs[1, 2].set_xlabel('Episode')
axs[1, 2].set_ylabel('Remaining Budget (out of 20)')
axs[1, 2].legend()
axs[1, 2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````
